# Desafio calculadora convesora de moedas

```python
date = input(f"Informe uma data (yyyy-mm-dd) ou escreva 'latest' se quiser ver os dados da data mais recente disponivel:\n")
base = input("codigo moeda a ser convertida: ")
curr = input("converter para: ")
qtd = float(input("Quanto {} gostaria de converter?: ".format(base)))

url = baseUrl + "/" + date + "?base=" + base + "&symbols=" + curr
response = requests.get(url)

if(response.ok is False):
    print("\nError {}:".format(response.status_code))
    print(response.text)

else:
    data = response.json()
    rate = data['rates'][curr]

    resultado = qtd*rate

    print("\n {0} {1} foi equivalente a {2} {3}, se baseando na cotação de {4}".format(base, qtd, curr, resultado, data['date']))
```
