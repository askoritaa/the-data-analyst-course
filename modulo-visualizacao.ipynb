Vamos usar matpllotlib e seaborn

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

usaremos o dataset dos carros disponibilizado pelo curso

```python
df = pd.read_csv("bar_chart_data.csv")
df
```

```python
plt.figure(figsize=(9,6)) 
plt.bar( x= df["Brand"],
        height= df["Cars Listings"],
        color = "orange")
plt.xticks(rotation = 45)
plt.show()
```

```python
df_pie_chart = pd.read_csv("pie_chart_data.csv")
df_pie_chart
```

```python
plt.figure(figsize=(10,8))
plt.pie(df_pie_chart['Number of Cars'],
        labels= df_pie_chart['Engine Fuel Type'].values,
        autopct= '%.2f%%',
        textprops= {
            'size': 'x-large',
            'fontweight': 'bold',
            'rotation': 'vertical',
            'color': 'w'
        })
plt.legend()
plt.title("Cars by Engine fuel Type", fontsize = 18, fontweight = 'bold')
plt.show()
```
